```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

# Define our text
text = "computers process data quickly"
words = text.split()
vocab = list(sorted(set(words)))

word_to_vec = {word: np.eye(len(vocab))[i] for i, word in enumerate(vocab)}

X = []
y = []
sequence = []

for word in words:
    sequence.append(word_to_vec[word])
    if len(sequence) > 1:
        X.append(np.array(sequence[:-1]))
        y.append(sequence[-1])

max_len = max(len(seq) for seq in X)
X_padded = np.array([np.pad(seq, ((max_len - len(seq), 0), (0, 0))) for seq in X])
y = np.array(y)

model = Sequential([
    SimpleRNN(32, input_shape=(max_len, len(vocab))),
    Dense(len(vocab), activation='softmax')
])
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.fit(X_padded, y, epochs=200, verbose=0)

def predict_next_word(input_sequence):
    seq_vectors = [word_to_vec[word] for word in input_sequence.split()]
    padded = np.pad(seq_vectors, ((max_len - len(seq_vectors), 0), (0, 0)))
    pred = model.predict(np.array([padded]), verbose=0)
    return vocab[np.argmax(pred)]

test_sequence = "computers process data"
prediction = predict_next_word(test_sequence)
print(f"Predicted next word after '{test_sequence}': {prediction}")
```
